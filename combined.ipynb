```python
for k,d in enumerate([fhc_beamline_unisim_cov, rhc_beamline_unisim_cov, beamline_unisim_cov]): 
    
    cov = [ [0]*(len(bins)-1) for x in range(len(bins)-1) ]
    frac_cov = [ [0]*(len(bins)-1) for x in range(len(bins)-1) ]
    cor = [ [0]*(len(bins)-1) for x in range(len(bins)-1) ]
    
    for variation in d.keys(): 

        for i in range(len(bins)-1): 
            for j in range(len(bins)-1):

                cov[i][j] += d[variation]['cov'][i][j]
                frac_cov[i][j] += d[variation]['frac_cov'][i][j] 


    for i in range(len(bins)-1): 
        for j in range(len(bins)-1):

            if np.sqrt(cov[i][i])*np.sqrt(cov[j][j]) != 0: 
                    cor[i][j] = cov[i][j] / (np.sqrt(cov[i][i])*np.sqrt(cov[j][j]))
        
    if k==0: 
        
        fhc_beamline_unisim_dict = {
            'cov' : cov, 
            'frac_cov' : frac_cov,
            'cor' : cor,
            'fractional_uncertainty' : np.sqrt(np.diag(frac_cov))
        } 
        
        
    elif k==1: 
        rhc_beamline_unisim_dict = {
            'cov' : cov, 
            'frac_cov' : frac_cov,
            'cor' : cor,
            'fractional_uncertainty' : np.sqrt(np.diag(frac_cov))
        } 
        
    elif k==2: 
        beamline_unisim_dict = {
            'cov' : cov, 
            'frac_cov' : frac_cov,
            'cor' : cor,
            'fractional_uncertainty' : np.sqrt(np.diag(frac_cov))
        } 
```

```python
# This comes from fakedata, and replaces beam-on data for closure test and NuWro. 
if isClosureTest or isNuWro:
    fhc_dict['beamon_counts'] = fhc_dict['evt_rate_nuwro_ppfx']
    rhc_dict['beamon_counts'] = rhc_dict['evt_rate_nuwro_ppfx']

if isClosureTest: 
    if isNuWro: 
        if ISRUN3:
            nuwro_full_sumw2 = rhc_dict['nuwro_full_sumw2'] # nuwro stat error
        
        else:
            nuwro_full_sumw2 = fhc_dict['nuwro_full_sumw2'] # nuwro stat error
            
    # This one is for when it's just the closure test (no NuWro)
    mc_signal_sumw2 = [a+b for a,b in zip(fhc_dict['mc_signal_sumw2'], rhc_dict['mc_signal_sumw2'])] # mc stat error
    
else: 
    cv_bkgd = [a+b for a,b in zip(fhc_dict['cv_bkgd'], rhc_dict['cv_bkgd'])]
    beamon_counts = [a+b for a,b in zip(fhc_dict['beamon_full_evt_rate'], rhc_dict['beamon_full_evt_rate'])] # data (background-subtracted)
    print(beamon_counts)

mc_bkgd_sumw2 = [a+b for a,b in zip(fhc_dict['mc_bkgd_sumw2'], rhc_dict['mc_bkgd_sumw2'])]
ext_sumw2 = [a+b for a,b in zip(fhc_dict['ext_sumw2'], rhc_dict['ext_sumw2'])]
evt_rate = [a+b for a,b in zip(fhc_dict['full_evt_rate'], rhc_dict['full_evt_rate'])]
```

```python
if isClosureTest: 
    
    for i in range(len(bins)-1):
        
        if isNuWro: 

            if ISRUN3:
                rhc_frac_stat_cov[i][i] = (nuwro_full_sumw2[i] + rhc_dict['mc_bkgd_sumw2'][i] + rhc_dict['ext_sumw2'][i]) / (rhc_dict['evt_rate_nuwro_ppfx'][i]**2)
                rhc_stat_percent_error.append( np.sqrt( rhc_frac_stat_cov[i][i]) ) 
            
            else:
                fhc_frac_stat_cov[i][i] = (nuwro_full_sumw2[i] + fhc_dict['mc_bkgd_sumw2'][i] + fhc_dict['ext_sumw2'][i]) / (fhc_dict['evt_rate_nuwro_ppfx'][i]**2) if fhc_dict['evt_rate_nuwro_ppfx'][i] != 0 else 0
                fhc_stat_percent_error.append( np.sqrt( fhc_frac_stat_cov[i][i]) ) 
        
        else: 
            stat_cov[i][i] = (mc_signal_sumw2[i] + mc_bkgd_sumw2[i] + ext_sumw2[i])

            frac_stat_cov[i][i] = stat_cov[i][i] / (evt_rate[i]**2) if evt_rate[i] != 0 else 0

            # this is OK -- just take the sum of the weights squared of the full event rate for GENIE closure test 
            fhc_frac_stat_cov[i][i] = ( fhc_dict['mc_signal_sumw2'][i] + fhc_dict['mc_bkgd_sumw2'][i] + fhc_dict['ext_sumw2'][i]) / (fhc_dict['evt_rate'][i]**2) if fhc_dict['evt_rate'][i] != 0 else 0
            rhc_frac_stat_cov[i][i] = ( rhc_dict['mc_signal_sumw2'][i] + rhc_dict['mc_bkgd_sumw2'][i] + rhc_dict['ext_sumw2'][i]) / (rhc_dict['evt_rate'][i]**2) if rhc_dict['evt_rate'][i] != 0 else 0

            fhc_stat_percent_error.append( np.sqrt( fhc_frac_stat_cov[i][i]) ) 
            rhc_stat_percent_error.append(np.sqrt(rhc_frac_stat_cov[i][i])) 
            stat_percent_error.append(np.sqrt(frac_stat_cov[i][i])) 
        
else: 
    
    # use the full event rates - before background subtraction
    delta_n_data2 = [a+b for a,b in zip(beamon_counts,cv_bkgd)] 
    delta_n_data2_fhc = [a+b for a,b in zip(fhc_dict['beamon_full_evt_rate'],fhc_dict['cv_bkgd'])]
    delta_n_data2_rhc = [a+b for a,b in zip(rhc_dict['beamon_full_evt_rate'],rhc_dict['cv_bkgd'])]
    
    for i in range(len(bins)-1):
    
        stat_cov[i][i] = (delta_n_data2[i] + mc_bkgd_sumw2[i] + ext_sumw2[i])
        mc_stat_cov[i][i] = mc_bkgd_sumw2[i]
        ext_stat_cov[i][i] = ext_sumw2[i]
        beamon_stat_cov[i][i] = delta_n_data2[i]

        mc_frac_stat_cov[i][i] = mc_stat_cov[i][i] / (evt_rate[i]**2) if evt_rate[i] != 0 else 0
        ext_frac_stat_cov[i][i] = ext_stat_cov[i][i] / (evt_rate[i]**2) if evt_rate[i] != 0 else 0
        beamon_frac_stat_cov[i][i] = beamon_stat_cov[i][i] / (evt_rate[i]**2) if evt_rate[i] != 0 else 0

        if delta_n_data2[i]!=0: 
            # frac_stat_cov[i][i] = stat_cov[i][i] / (delta_n_data2[i]**2)
            frac_stat_cov[i][i] = stat_cov[i][i] / (evt_rate[i]**2) if evt_rate[i] != 0 else 0

            fhc_denom = fhc_dict['beamon_full_evt_rate'][i]
            rhc_denom = rhc_dict['beamon_full_evt_rate'][i]
            fhc_frac_stat_cov[i][i] = ( delta_n_data2_fhc[i] + fhc_dict['mc_bkgd_sumw2'][i] + fhc_dict['ext_sumw2'][i]) / (fhc_denom**2) if fhc_denom != 0 else 0
            rhc_frac_stat_cov[i][i] = ( delta_n_data2_rhc[i] + rhc_dict['mc_bkgd_sumw2'][i] + rhc_dict['ext_sumw2'][i]) / (rhc_denom**2) if rhc_denom != 0 else 0

        fhc_stat_percent_error.append( np.sqrt( fhc_frac_stat_cov[i][i]) ) 
        rhc_stat_percent_error.append(np.sqrt(rhc_frac_stat_cov[i][i])) 
        stat_percent_error.append(np.sqrt(frac_stat_cov[i][i]))
```
